## Objectives and dataset
Explore data quality, distributions, numeric relationships, traffic and vehicle group differences, shipment status, and monthly patterns. The dataset spans Jan?Dec 2025 and has 12 fields. Distance is sampled from a clipped gamma-like range (1?55 km), weight from a clipped lognormal range (0.3?180 kg). Delivery duration combines distance, vehicle-specific assumed speed, traffic increments, a small weight term, and random noise. Cost combines distance, assumed vehicle rate, weight, and noise. Delay is assigned probabilistically using traffic and simulated duration. These assumptions create plausible variation, not validated operational behavior.

## Time trend and operational implications
Monthly mean duration ranged from 42.28 minutes (June) to 48.15 (February); this synthetic year does not encode a designed seasonal pattern, so month-to-month movement should be treated as sampling variation. In a real operation, stratify monitoring by distance, traffic, vehicle, and promised service level; examine cost per shipment and high-delay segments before capacity changes. ## Limitations and conclusion
This is synthetic cross-sectional data from a simple stipulated generator. It has no actual route network, promised delivery timestamps, service area definition, currency specification, dispatch events, vehicle capacity, or company provenance. Generated relationships are teaching examples, not evidence for decisions. EDA demonstrates how summaries, correlations, and group comparisons can reveal patterns for further investigation. Pillow charts were used because matplotlib is absent in this environment; the analysis script uses matplotlib automatically when installed.
